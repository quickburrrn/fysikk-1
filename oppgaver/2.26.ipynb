```python
from pylab import *

m = 10
g = 9.81
mu = 0.32
G = m*g
k = 0.1

def a(v):
    L = k*v**2
    sum_F = G - L
    aks = sum_F/m
    return aks

s = 0.0
v = 0.0
t = 0.0

# sluttpunkt
v_slutt = 0
t_slutt = 4.0
s_slutt = 100.0
a_slutt = 100

dt = 0.0001
s_verdier = [s]
v_verdier = [v]
t_verdier = [t]
kmt_verdier = [v*3.6]

while s < s_slutt:
    v = v + a(v)*dt
    s = s + v*dt
    t = t + dt

    t_verdier.append(t)
    v_verdier.append(v)
    kmt_verdier.append(v*3.6)
    s_verdier.append(s)

plot(t_verdier, kmt_verdier)
title("Far som funksjon av tid")
xlabel("$t$ / s")
ylabel("$v$ / (mk/T)")
grid()
show()

print("Sluttfart:", v, "m/s")
print("Slutttid:", t, "s")
```
