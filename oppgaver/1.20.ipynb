# 1.20 En ball starter fra ro og har konstant akselerasjon på 2,0m/s
### a) Lag en simulering som viser hvor langt ballen har gått etter 5,0s

```python
# Variabler
s = 0.0 # startposisjon, m
v = 0 # startfart, m/s
a = 2.0 # akselerasjon, m/s^2
t = 0.0 # starttid, s

t_slutt = 5.0 # sluttid, s
dt = 0.1 # deltatid, s

while t < t_slutt:
    v += a*dt
    s += v*dt
    t += dt

print(s)
```

### b) Sammenlign svaret fra oppgave a med en dirrekte utregning 

```python
s = (1/2) * a * t_slutt**2 
print(s)

# ser ut til å være 1.5 forskjell mellom simulering og dirrekte regning
```

### c) hvor langt kommer ballen hvis vi setter dt=0.01? Komenter svaret

```python
# Variabler
s = 0.0 # startposisjon, m
v = 0 # startfart, m/s
a = 2.0 # akselerasjon, m/s^2
t = 0.0 # starttid, s

t_slutt = 5.0 # sluttid, s
dt = 0.01 # deltatid, s

while t < t_slutt:
    v += a*dt
    s += v*dt
    t += dt

print(s)

# Svaret blir mer nøyaktig
```
