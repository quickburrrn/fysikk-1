```python
from pylab import *

m = 0.250
g = 9.81
mu = 0.3
G = m*g
N = G
R = mu*N

sum_F = -R
a = sum_F/m

s = 0.0
v = 2.97136
t = 0.0

# sluttpunkt
v_slutt = 0
t_slutt = 4.0

dt = 0.0001
s_verdier = [s]
v_verdier = [v]
t_verdier = [t]
 
while t < t_slutt:
    v = v + a*dt
    s = s + v*dt
    t = t + dt

    t_verdier.append(t)
    v_verdier.append(v)
    s_verdier.append(s)

plot(t_verdier, s_verdier)
title("Strekning som funksjon av tid")
xlabel("$t$ / s")
ylabel("$s$ / m")
grid()
show()

print("Sluttposisjon:", s, "m")
```
