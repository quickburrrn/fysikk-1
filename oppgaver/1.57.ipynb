Vi kaster en stein opp i lufta med startfart på 8,0m/s. Se bort fra luftmotstand

### ) Lag en simulering og finn posisjonen til steinenn ved tiden $t=2,0s$

```python
s = 0.0
v = 8.0
t = 0.0
a = -9.81

t_slutt = 2.0
dt = 0.001

while t < t_slutt:
    v += a*dt
    s += v*dt
    t += dt
# 
print(f"posisjonen til steinen ved tiden t=2.0 er {round(s, 1)}m")
```

### b) Finn posisjon med dirrekte utregning og sammenlign

$S=v_0t + at^2 \frac{1}{2}=8*2-9.81*2^2*\frac{1}{2}=-3,6m$

Vi ser at det er en 0.01 forskjell på a og b men fortsatt ganske likt

### c) Finn høyden på kastet ved hjelp av simulering og ved hjelp av rørselikningene

```python
s = 0.0
v = 8.0
t = 0.0
a = -9.81

t_slutt = 2.0
dt = 0.001

s_verdier = [s]

while t < t_slutt:
    v += a*dt
    s += v*dt
    t += dt
    s_verdier.append(s)
# 
print(f"Høyden på kastet er {round(max(s_verdier), 1)}m")
```

Hvis man skal gjøre det med ligning

$$
\begin{align}
s'(t) &= 0 \\
s'(t) &=v(t)=v_0+at \\
v_0+at &= 0 \rightarrow 8,0 m/s-9,81m/s^2 \cdot t = 0 \\
t &= 0,82s \\ \\
S &= 8,0s \cdot 0,82s + \frac{1}{2} \cdot -9,81 \cdot 0,82^2=3.3m
\end{align}
