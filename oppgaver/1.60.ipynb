Vi slepper en ball fra en 20m høy bygning slik at farten bli så stor at vi ikke kan se bort fra luftmotsand. Akselerasjonen til ballen er $a(v)=-9,81m/s^2 + 0,034m^(-1) \cdot v^2$

### a) Lag et program som simulerer bevegelsen, og som finner ut hvor lang tid ballen bruker før den treffer bakken

```python
s = 20
v = 0.0
t = 0.0

s_slutt = 0
dt = 0.001

def a(v):
    return -9.81 + 0.034 * v**2

while s > s_slutt:
    v += a(v)*dt
    s += v*dt
    t += dt

print(f"Det tar {round(t, 1)}s før ballen treffer bakken")
```
