# 1.22 Lag et program som simulerer en ball med startfart på 0m/s som er i størrelse med konstant akselerasjon lik 4,0m/s^2 i 5,0s
### Lag programmet som tegner strekningen som funksjon av tida

```python
import matplotlib.pyplot as plt

# Variabler
s = 0.0 # startposisjon, m
v = 0 # startfart, m/s
a = 4.0 # akselerasjon, m/s^2
t = 0.0 # starttid, s

t_slutt = 5.0 # sluttid, s
dt = 0.1 # deltatid, s
t_verider = [t]
s_verdier = [s]

while t < t_slutt:
    v += a*dt
    s += v*dt
    t += dt
    t_verider.append(t)
    s_verdier.append(s)

plt.plot(t_verider, s_verdier)
plt.title("Strekning som funksjon av tid")
plt.xlabel("$t$ (s)")
plt.ylabel("$s$ (m)")
plt.grid()
plt.show()
```
