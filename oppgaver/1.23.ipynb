### a) lag et program som kan skrive hvor langt tid løperen bruker på 100 m

```python
import matplotlib.pyplot as plt

s = 0.0 # startposisjon, m
v = 0.0 # startfart, m/s
t = 0.0 # starttid, s

s_slutt = 100 # sluttposisjon, m
dt = 0.01 
s_verdier = [s]
v_verdier = [v]
t_verider = [t]

def a(v):
    aks = -0.5*v + 6
    return aks

while s < s_slutt:
    v += a(v)*dt
    s += v*dt
    t += dt
    s_verdier.append(s)
    v_verdier.append(v)
    t_verider.append(t)

plt.plot(s_verdier, v_verdier)
plt.title("Fart som funksjon av posisjon")
plt.xlabel("$s$ (m)")
plt.ylabel("$v$ (m/s)")
plt.grid()
plt.show()

# a)
print("løperen brukte ", {t}, "$$")
```

```python
plt.plot(t_verider, s_verdier)
plt.title("Strekning som funksjon av tid")
plt.xlabel("$t$ (s)")
plt.ylabel("$s$ (m)")
plt.grid()
plt.show()
```
