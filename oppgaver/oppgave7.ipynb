# Oppgave 7 kapitteltest
En kule blir skutt av et luftgeværkøp som er 40cm langt. Akselerasjonen til kule i $m/s^2$ er gitt ved $a(s)=300/(s+0,01)$, der $s$ er lengda kula har flyttet seg i geværløået i meter.

```python
s = 0.0
v = 0.0
t = 0.0

s_slutt = 0.4
dt = 0.0001

def a(s):
    return 300/(s+0.01)

while s < s_slutt:
    v += a(s)*dt
    s += v*dt
    t += dt

print(f"kula forlater geværet med en fart på {round(v)}m/s")
```
