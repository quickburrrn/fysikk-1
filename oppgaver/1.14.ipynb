En bil kjører fra et lyskryss. Akselerasjonen til bilen de første ti sekundene er $a(t)=-0,40m/s^3 \cdot 4,0 m/s^2$
### a) Lag en simulering som viser farten til bilen som funksjonen av tiden

```python
import matplotlib.pyplot as plt

s = 0.0
v = 0.0
t = 0.0

s_slutt = 10
dt = 0.01
t_verdier = [t]
s_verdier = [s]
v_verdier = [v]

def a(t):
    aks = -0.4 * t + 4.0
    return aks

while t < s_slutt:
    v += a(t)*dt
    s += v*dt
    t += dt

    t_verdier.append(t)
    s_verdier.append(s)
    v_verdier.append(v)

plt.plot(t_verdier, v_verdier)
plt.title("Strekning som funksjon av tida")
plt.xlabel("$t$ (s)")
plt.ylabel("$s$ (m)")
plt.grid()
plt.show()

#b)
print("bilen kommer",s,"m")
```
