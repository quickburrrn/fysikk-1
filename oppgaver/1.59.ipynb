En rakett har en akselerasjon på $a(t)=0,50 m/s^3 \cdot t + 8,0 m/s^2$
### a og b) Simler hastigheten til raketten de første ti sekundene så tegn akselerasjonen, farten og posisjonen til raketten som funksjon av tiden

```python
import matplotlib.pyplot as plt

s = 0.0
v = 0.0
t = 0.0

t_slutt = 10
dt = 0.01
s_verdier = [s]
v_verdier = [v]
t_verdier = [t]


def a(t):
    return 0.5 * t + 8

a_verdier = [a(t)]

while t < t_slutt:
    v += a(t)*dt
    s += v*dt
    t += dt
    a_verdier.append(a(t))
    v_verdier.append(v)
    s_verdier.append(s)
    t_verdier.append(t)


# a)
print(f"Raketen har en fart på {round(s*0.001, 2)}km/s etter 10s")

plt.plot(t_verdier, a_verdier)
plt.title("Akselerasjon som funksjon at tiden")
plt.xlabel("$t$ (s)")
plt.ylabel("$v$ (m/s)")
plt.grid()
plt.show()

plt.plot(t_verdier, v_verdier)
plt.title("Fart som funksjon at tiden")
plt.xlabel("$t$ (s)")
plt.ylabel("$v$ (m/s)")
plt.grid()
plt.show()

plt.plot(t_verdier, s_verdier)
plt.title("Posisjon som funksjon at tiden")
plt.xlabel("$t$ (s)")
plt.ylabel("$s$ (m)")
plt.grid()
plt.show()
```
